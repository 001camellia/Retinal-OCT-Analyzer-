```python
import os
import zipfile
from PIL import Image

def crop_left_42_percent(image):
    """
    裁剪图片，保留右边58%，去掉左边42%，同时保留顶部81.55%，去掉底部18.45%
    :param image: 图片对象
    :return: 裁剪后的图片对象
    """
    width, height = image.size
    # 计算左侧裁剪点
    left = int(width * 0.394)
    # 计算底部裁剪点
    bottom = int(height * 0.8155)
    # 定义裁剪区域
    box = (left, 0, width, bottom)
    # 执行裁剪
    cropped_img = image.crop(box)
    return cropped_img

def process_images_in_directory(directory):
    """
    处理指定目录下的所有图片文件，包括子目录
    :param directory: 目录路径
    """
    processed_files_count = 0  # 添加计数器
    skipped_files = []  # 记录被跳过的文件
    print(f"开始处理目录: {directory}")
    for root, dirs, files in os.walk(directory):
        for file in files:
            file_path = os.path.join(root, file)
            if file.endswith(('.png', '.jpg', '.jpeg')):
                print(f"处理图片: {file_path}")
                try:
                    with Image.open(file_path) as img:
                        cropped_img = crop_left_42_percent(img)
                        cropped_img.save(file_path)  # 保存裁剪后的图片，覆盖原文件
                    processed_files_count += 1  # 增加计数器
                except PermissionError as e:
                    print(f"权限错误: 无法访问文件 {file_path} - {e}")
                    skipped_files.append((file_path, "权限错误"))
                except Exception as e:
                    print(f"处理文件 {file_path} 时发生错误: {e}")
                    skipped_files.append((file_path, "处理错误"))
            else:
                print(f"跳过非图片文件: {file_path}")
                skipped_files.append((file_path, "非图片文件"))
    print(f"处理完成，共处理了 {processed_files_count} 个文件")
    if skipped_files:
        print("以下文件被跳过:")
        for file, reason in skipped_files:
            print(f"{file} - 原因: {reason}")
    return processed_files_count, skipped_files  # 返回处理的文件数量和被跳过的文件列表

def zip_directory(src_dir, output_zip):
    """
    压缩指定目录到一个zip文件
    :param src_dir: 要压缩的目录
    :param output_zip: 输出的zip文件路径
    """
    with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(src_dir):
            for file in files:
                file_path = os.path.join(root, file)
                arcname = os.path.relpath(file_path, src_dir)
                zipf.write(file_path, arcname)

if __name__ == '__main__':
    # 指定输入文件夹的绝对路径
    input_folder = r'D:\各类课题\眼科天池杯\train\OCT'
    
    total_processed_files = 0  # 总处理文件计数器
    all_skipped_files = []  # 记录所有被跳过的文件

    # 遍历包含 '-' 的目录
    for filename in os.listdir(input_folder):
        if '-' in filename:
            input_path = os.path.join(input_folder, filename)
            if os.path.isdir(input_path):
                print(f"处理目录: {input_path}")
                processed_count, skipped_files = process_images_in_directory(input_path)
                total_processed_files += processed_count
                all_skipped_files.extend(skipped_files)
            else:
                print(f"跳过非目录项: {filename}")
        else:
            print(f"跳过不含 '-' 的目录项: {filename}")
    
    if total_processed_files != 730:
        print(f"警告：处理的文件数量 ({total_processed_files}) 不符合预期 (730)")
    else:
        print("所有文件均已被正确处理")
    
    # 输出所有被跳过的文件
    if all_skipped_files:
        print("所有被跳过的文件:")
        for file, reason in all_skipped_files:
            print(f"{file} - 原因: {reason}")
    else:
        print("没有文件被跳过")

    # 压缩处理后的 OCT 文件夹
    output_zip = r'D:\各类课题\眼科天池杯\train\OCT.zip'
    zip_directory(input_folder, output_zip)
    
    print("处理完成，图片已裁剪并保存，文件夹已压缩")
```
